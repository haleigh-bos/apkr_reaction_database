```python
# import libraries

from rdkit import Chem
from rdkit.Chem.Draw import IPythonConsole
from rdkit.Chem import Draw
IPythonConsole.ipython_usesSVG=True #< set this to False if you want PNGs instead of SVGs

import pandas as pd
```

```python
# Converts SMILES into images (.png format)
def make_png(ref_sheet, target, list):
    png_locs = '../2.png_files/'

    mol_num = 1
    mol_list = []
    mol_list_2 = []
    mol_list_3 = []
    
    for smiles in list:
        mol = Chem.MolFromSmiles(smiles)
        if mol_num < 50:
            mol_list.append(mol)
        elif 50 < mol_num < 100:
            mol_list_2.append(mol)
        else:
            mol_list_3.append(mol)
        mol_num += 1


    if len(mol_list_2) != 0:
        grid_image = Draw.MolsToGridImage(mol_list, molsPerRow=4, subImgSize=(200,200), returnPNG=False)
        grid_image.save(f'{png_locs}/{target}_1_{ref_sheet}.png')
    else:
        grid_image = Draw.MolsToGridImage(mol_list, molsPerRow=4, subImgSize=(200,200), returnPNG=False)
        grid_image.save(f'{png_locs}/{target}_{ref_sheet}.png')

    if len(mol_list_2) != 0:
        grid_image_2 = Draw.MolsToGridImage(mol_list_2, molsPerRow=4, subImgSize=(200,200), returnPNG=False)
        grid_image_2.save(f'{png_locs}/{target}_2_{ref_sheet}.png')

    if len(mol_list_3) != 0:
        grid_image_3 = Draw.MolsToGridImage(mol_list_3, molsPerRow=4, subImgSize=(200,200), returnPNG=False)
        grid_image_3.save(f'{png_locs}/{target}_3_{ref_sheet}.png')

   
```

```python
# Load data from excel sheets into neat lists
sheet_locs = "../0.database_excels/"
excel_sheets = ["Burrows_JACS_2017_08.07.2026", "Kim_JOC_2008_08.07.2026", "Qi_ACS-Catal_2024_08.07.2026"]
substrate_list = []
ligand_list = []
product_list = []
byproduct_list = []

track_num = 1

# Systematically make lists of compounds
for file in excel_sheets: 
    file_loc = f'{sheet_locs}{file}.xlsx'
    excel_df = pd.read_excel(file_loc)

    substrates = excel_df['Substrate SMILES'].tolist()
    ligands = excel_df['Ligand SMILES'].tolist()
    products = excel_df['Product SMILES'].tolist()
    byproducts = excel_df['Byproduct SMILES'].tolist()

    # Checks for duplicates in the lists, only adds the SMILES that are not already represented
    inter_substrates = []
    for smiles in substrates:
        if smiles not in inter_substrates:
            inter_substrates.append(smiles)

    inter_ligands = []
    for smiles in ligands:
        if smiles not in ligand_list:
            inter_ligands.append(smiles)

    inter_products = []
    for smiles in products:
        if smiles not in inter_products:
            inter_products.append(smiles)

    # inter_byproducts = []
    # for smiles in byproducts:
    #     if smiles not in inter_byproducts:
    #         inter_byproducts.append(smiles)
    
    make_png(file, 'substrates', inter_substrates)
    substrate_list.append(inter_substrates)
    # make_png(file, 'ligands', inter_ligands)
    ligand_list.append(inter_ligands)
    make_png(file, 'products', inter_products)
    product_list.append(inter_products)
    # make_png(file, 'byproducts', inter_byproducts)

    print(f'Appended lists with {file} data. Substrates: {len(inter_substrates)}.')
    track_num += 1

print(f'Total substrates after {track_num} files: {len(substrate_list)}')
print(f'Total ligands after {track_num} files: {len(ligand_list)}')
print(f'Total products after {track_num} files: {len(product_list)}')
print(f'Total byproducts after {track_num} files: {len(byproduct_list)}')

print(substrate_list)
print(ligand_list)
```
